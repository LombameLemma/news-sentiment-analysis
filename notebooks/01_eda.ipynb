```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Show plots inside notebook
%matplotlib inline
news = pd.read_csv("../data/raw/news.csv")

news.head()
news.shape
news.columns
news.info()
news.isnull().sum()
news.describe(include='all')
news['headline_length'] = news['headline'].astype(str).apply(len)

news[['headline', 'headline_length']].head()
news['headline_length'].describe()
plt.figure(figsize=(10,6))

plt.hist(news['headline_length'], bins=30)

plt.title("Headline Length Distribution")
plt.xlabel("Headline Length")
plt.ylabel("Frequency")

plt.show()
top_publishers = news['publisher'].value_counts().head(10)

top_publishers
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_publishers.values,
    y=top_publishers.index
)

plt.title("Top 10 Publishers")
plt.xlabel("Number of Articles")
plt.ylabel("Publisher")

plt.show()
top_stocks = news['stock'].value_counts().head(10)

top_stocks

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_stocks.index,
    y=top_stocks.values
)

plt.title("Top 10 Most Mentioned Stocks")
plt.xlabel("Stock")
plt.ylabel("Article Count")

plt.show()
news['date'] = pd.to_datetime(
    news['date'],
    errors='coerce'
)

news['date'].head()
news['day'] = news['date'].dt.date

news[['date', 'day']].head()
daily_articles = news.groupby('day').size()

daily_articles.head()
plt.figure(figsize=(14,6))

daily_articles.plot()

plt.title("Articles Published Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Articles")

plt.show()
news['weekday'] = news['date'].dt.day_name()

news['weekday'].value_counts()
weekday_counts = news['weekday'].value_counts()

plt.figure(figsize=(10,6))

sns.barplot(
    x=weekday_counts.index,
    y=weekday_counts.values
)

plt.title("Articles by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

## Key Insights

- Most headlines contain between 50–100 characters.
- Reuters is the dominant publisher.
- AAPL and TSLA appear most frequently.
- News volume increased significantly during earnings periods.
