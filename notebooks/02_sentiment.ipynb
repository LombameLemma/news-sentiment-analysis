```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
news = pd.read_csv("../data/raw/news.csv")

news.head()
analyzer = SentimentIntensityAnalyzer()
def get_sentiment(text):

    score = analyzer.polarity_scores(str(text))

    return score['compound']
    news['sentiment_score'] = news['headline'].apply(get_sentiment)

news[['headline', 'sentiment_score']].head()
def sentiment_label(score):

    if score > 0.05:
        return 'Positive'

    elif score < -0.05:
        return 'Negative'

    else:
        return 'Neutral'
        news['sentiment'] = news['sentiment_score'].apply(sentiment_label)

news[['headline', 'sentiment_score', 'sentiment']].head()
news['sentiment'].value_counts()
plt.figure(figsize=(8,6))

sns.countplot(
    x='sentiment',
    data=news
)

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
plt.figure(figsize=(10,6))

plt.hist(
    news['sentiment_score'],
    bins=30
)

plt.title("Sentiment Score Distribution")
plt.xlabel("Sentiment Score")
plt.ylabel("Frequency")

plt.show()
stock_sentiment = news.groupby('stock')['sentiment_score'].mean()

stock_sentiment.head()
top_positive = stock_sentiment.sort_values(ascending=False).head(10)

top_positive

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_positive.index,
    y=top_positive.values
)

plt.title("Top 10 Positive Sentiment Stocks")
plt.xlabel("Stock")
plt.ylabel("Average Sentiment")

plt.show()
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_positive.index,
    y=top_positive.values
)

plt.title("Top 10 Positive Sentiment Stocks")
plt.xlabel("Stock")
plt.ylabel("Average Sentiment")

plt.show()
top_negative = stock_sentiment.sort_values().head(10)

top_negative
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_negative.index,
    y=top_negative.values
)

plt.title("Top 10 Negative Sentiment Stocks")
plt.xlabel("Stock")
plt.ylabel("Average Sentiment")

plt.show()
news['date'] = pd.to_datetime(
    news['date'],
    errors='coerce'
)

news['day'] = news['date'].dt.date
daily_sentiment = news.groupby('day')['sentiment_score'].mean()

daily_sentiment.head()
plt.figure(figsize=(14,6))

daily_sentiment.plot()

plt.title("Daily Average Sentiment")
plt.xlabel("Date")
plt.ylabel("Average Sentiment Score")

plt.show()
news.sort_values(
    by='sentiment_score',
    ascending=False
)[['headline', 'sentiment_score']].head(10)
news.sort_values(
    by='sentiment_score'
)[['headline', 'sentiment_score']].head(10)
```

## Sentiment Analysis Insights

- Most financial headlines are neutral.
- Positive sentiment dominates technology-related stocks.
- Negative sentiment spikes during market uncertainty.
- Some stocks consistently receive stronger emotional coverage.
