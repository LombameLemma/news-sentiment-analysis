```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ta.trend import SMAIndicator, MACD
from ta.momentum import RSIIndicator
stocks = pd.read_csv("../data/raw/stock_prices.csv")

stocks.head()
stocks.info()
stocks['Date'] = pd.to_datetime(stocks['Date'])

stocks.head()
stocks = stocks.sort_values('Date')
stocks['Daily_Return'] = stocks['Adj Close'].pct_change()

stocks[['Date', 'Adj Close', 'Daily_Return']].head()
plt.figure(figsize=(14,6))

plt.plot(
    stocks['Date'],
    stocks['Daily_Return']
)

plt.title("Daily Stock Returns")
plt.xlabel("Date")
plt.ylabel("Daily Return")

plt.show()
sma_10 = SMAIndicator(
    close=stocks['Adj Close'],
    window=10
)

stocks['SMA_10'] = sma_10.sma_indicator()
sma_20 = SMAIndicator(
    close=stocks['Adj Close'],
    window=20
)

stocks['SMA_20'] = sma_20.sma_indicator()
plt.figure(figsize=(14,6))

plt.plot(
    stocks['Date'],
    stocks['Adj Close'],
    label='Adj Close'
)

plt.plot(
    stocks['Date'],
    stocks['SMA_10'],
    label='SMA 10'
)

plt.plot(
    stocks['Date'],
    stocks['SMA_20'],
    label='SMA 20'
)

plt.title("Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()

plt.show()
rsi = RSIIndicator(
    close=stocks['Adj Close'],
    window=14
)

stocks['RSI'] = rsi.rsi()
plt.figure(figsize=(14,6))

plt.plot(
    stocks['Date'],
    stocks['RSI']
)

plt.axhline(70, linestyle='--')
plt.axhline(30, linestyle='--')

plt.title("RSI Indicator")
plt.xlabel("Date")
plt.ylabel("RSI")

plt.show()
macd = MACD(close=stocks['Adj Close'])

stocks['MACD'] = macd.macd()
stocks['MACD_signal'] = macd.macd_signal()
plt.figure(figsize=(14,6))

plt.plot(
    stocks['Date'],
    stocks['MACD'],
    label='MACD'
)

plt.plot(
    stocks['Date'],
    stocks['MACD_signal'],
    label='Signal Line'
)

plt.title("MACD Indicator")
plt.xlabel("Date")
plt.ylabel("MACD")

plt.legend()

plt.show()
stocks['Volatility'] = (
    stocks['Daily_Return']
    .rolling(window=10)
    .std()
)
plt.figure(figsize=(14,6))

plt.plot(
    stocks['Date'],
    stocks['Volatility']
)

plt.title("Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")

plt.show()
stocks[
    [
        'Daily_Return',
        'SMA_10',
        'SMA_20',
        'RSI',
        'MACD'
    ]
].describe()
plt.figure(figsize=(10,6))

sns.heatmap(
    stocks[
        [
            'Daily_Return',
            'RSI',
            'MACD',
            'Volatility'
        ]
    ].corr(),
    annot=True
)

plt.title("Technical Indicators Correlation")

plt.show()
```

## Technical Indicator Insights

- Moving averages smooth short-term price fluctuations.
- RSI identifies potential overbought and oversold conditions.
- MACD captures momentum shifts.
- Volatility spikes during large market movements.
